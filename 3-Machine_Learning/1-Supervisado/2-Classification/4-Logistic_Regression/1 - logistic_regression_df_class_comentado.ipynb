## Creamos el modelo

```python
X = np.array(dataframe.drop(columns=['clase']))
y = np.array(dataframe['clase'])
```

```python
X = dataframe.drop("clase",axis=1).values
y = dataframe["clase"].values
```

```python
X
```

```python
y
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression(max_iter=1000)       ##max iter: limite de operaciones para llegar al minimo, muchas veces no se llega
model.fit(X, y)
```

```python
predictions = model.predict(X)      ##le entrena con todos los datos, no queda test. Habia pocos datos
print(predictions)
```

```python
predicions_proba = model.predict_proba(X)       ##probabilidades: 0.22 windows, 0.09 Linux, 0.69 Mac, para cada usuario.
print(np.round(np.array(predicions_proba), 2))  ##redondeo a 2 decimales
#predicions_proba
```

```python
model.score(X, y)       ##F1, es asi asi, 0.77
```

```python
model.classes_          ##clases que puede asignar, 0, 1 y 2.
```

# Validación del Modelo

```python
validation_size = 0.20
seed = 7
X_train, X_test, Y_train, Y_test = model_selection.train_test_split(X,
                                                                    y,
                                                                    test_size=validation_size,
                                                                    random_state=seed)
```

```python
name='Logistic Regression'              ##cross validation
kfold = model_selection.KFold(n_splits=10)          #Parte los datos en 10 trozos para usar validación cruzada / cross validation
cv_results = model_selection.cross_val_score(model, X_train, Y_train, cv=kfold, scoring='accuracy')

msg = "%s: %f (%f)" % (name, cv_results.mean(), cv_results.std())
print(cv_results)
print(msg)   ##el modelo acierta el 72% de las veces           ##es una metrica mas correcta que probarlo solo una vez, poruqe lo ha hecho 10 veces y sale un accuracy diferente cada vez
```

```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train, Y_train)
```

```python
predictions = model.predict(X_test)
print(accuracy_score(Y_test, predictions))      ##mejor accuracy en test que en train.
predictions
```

```python
acierto = accuracy_score(Y_test, predictions)

error = 1 - acierto
print("Acierto:", round(acierto*100, 2), "%")
print("Error:", round(error*100, 2), "%")
```
