```python
import pandas as pd
import numpy as np
import matplotlib as mpl
import seaborn as sns

df = pd.read_csv('train.csv')
df.head()
```

```python
df["Weight"] = df["Weight"].str.replace("kg", "").astype(float)     ## limpiar columna numerica
df.head()
```

```python
df["Ram"] = df["Ram"].str.replace("GB", "").astype(int)         ## limpiar columna numerica
df.head()
```

```python
df["Resolution"] = df["ScreenResolution"].str.extract(r"(\d+x\d+)")     ## procesar screenresolution
df[["Width","Height"]] = df["Resolution"].str.split("x", expand=True).astype(int)
df.head()
```

```python
df["Cpu_freq"] = df["Cpu"].str.extract(r"(\d+\.\d+)").astype(float)     ## proesar CPU
df["Cpu_brand"] = df["Cpu"].str.split().str[0]
df["Cpu_model"] = df["Cpu"].str.extract(r"(i3|i5|i7|i9|Ryzen\s?\d)")
df.head()
```

```python
df["Gpu_brand"] = df["Gpu"].str.split().str[0]      ## procesar GPU
df.head()
```

```python
df = pd.get_dummies(df, drop_first=True)
```

```python
import re               ## procesar memory

def parse_memory(mem):
    mem = mem.replace(" ", "")
    ssd = hdd = flash = hybrid = 0

    parts = mem.split("+")
    for p in parts:
        if "SSD" in p:
            ssd += int(re.findall(r"(\d+)", p)[0])
        elif "HDD" in p:
            hdd += int(re.findall(r"(\d+)", p)[0])
        elif "Flash" in p:
            flash += int(re.findall(r"(\d+)", p)[0])
        elif "Hybrid" in p:
            hybrid += int(re.findall(r"(\d+)", p)[0])
    return pd.Series([ssd, hdd, flash, hybrid])

df[["SSD","HDD","Flash","Hybrid"]] = df["Memory"].apply(parse_memory)
df.head()
```

```python
df = pd.get_dummies(df, columns=[                               ## one hot encoding, codificar categoricas en columnas si o no
    "Company","TypeName","OpSys","Cpu_brand","Gpu_brand"
], drop_first=True)
df.head()
```

```python
df = df.drop(columns=[                          ## eliminar columnas inutiles
    "id","laptop_ID","Product","ScreenResolution","Cpu","Gpu","Memory"
])
df.head()
```

```python
X = df.drop('Price_euros', axis=1)          ## separar x e y
y = df['Price_euros']
```

```python
from sklearn.preprocessing import StandardScaler                ## escalar variables numericas

num_cols = ["Inches","Weight","Ram","Width","Height","Cpu_freq","SSD","HDD","Flash","Hybrid"]
df[num_cols] = StandardScaler().fit_transform(df[num_cols])
df.head()
```

```python
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_iris

# Datos
X, y = load_iris(return_X_y=True)

# Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Modelo SVM
model = SVC(kernel='linear')  # kernel lineal
model.fit(X_train, y_train)

# Evaluación
accuracy = model.score(X_test, y_test)
print("Accuracy:", accuracy)
```

```python
from sklearn.svm import SVC
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_iris

X, y = load_iris(return_X_y=True)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = SVC()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
```
