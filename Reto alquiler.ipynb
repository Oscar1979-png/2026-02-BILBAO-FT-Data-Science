## Limpieza, eliminamos registros no reales

```python
df_clean = df[
    (df['rent']== 1)&
    (df['price'] > 0) & (df['price'] < 5000) &
    (df['surface'] > 0) & (df['surface'] < 300) &
    (df['bedrooms'] > 0) & (df['bedrooms'] <= 10) &
    (df['restrooms'] > 0) & (df['restrooms'] <= 5)
].copy()

print(f"Filas tras limpieza: {len(df_clean)}")
print(f"Filas eliminadas: {len(df) - len(df_clean)}")
```

```python
df['features_list'] = df['features'].str.split(',\s*')
```

```python
df['m2'] = df['features'].str.extract(r'(\d+)\s*m2').astype(float)
df['habitaciones'] = df['features'].str.extract(r'(\d+)\s*hab').astype(float)
df['baños'] = df['features'].str.extract(r'(\d+)\s*bañ').astype(float)
```

```python
amenities = ['Terraza', 'Ascensor', 'Piscina', 'Calefaccion']

for a in amenities:
    df[a] = df['features'].str.contains(a, case=False, na=False).astype(int)
```
