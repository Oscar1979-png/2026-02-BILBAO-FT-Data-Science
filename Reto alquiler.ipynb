## Limpieza, eliminamos registros no reales

```python
df_clean = df[
    (df['rent']== 1)&
    (df['price'] > 0) & (df['price'] < 5000) &
    (df['surface'] > 0) & (df['surface'] < 300) &
    (df['bedrooms'] > 0) & (df['bedrooms'] <= 10) &
    (df['restrooms'] > 0) & (df['restrooms'] <= 10)
].copy()

print(f"Filas tras limpieza: {len(df_clean)}")
print(f"Filas eliminadas: {len(df) - len(df_clean)}")
```
