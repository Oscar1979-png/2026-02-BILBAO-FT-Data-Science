```python
## checking outliers

numeric_cols = mobile.select_dtypes(include=['number', 'float', 'int']).columns.tolist()
z_outliers = pd.DataFrame(index=mobile.index)

for col in numeric_cols:
    mean = mobile[col].mean()
    std  = mobile[col].std(ddof=0)

    # avoid division by zero
    if std == 0:
        z_outliers[col] = False
        continue

    z_scores = (mobile[col] - mean) / std
    z_outliers[col] = (z_scores.abs() > 3)

z_outlier_summary = z_outliers.sum().sort_values(ascending=False)
z_outlier_summary
```

```python
# =========================
# 0. Librerías y configuración
# =========================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (8, 5)

# =========================
# 1. Carga e inspección inicial
# =========================
df = pd.read_csv("SMA.csv")  # ajusta el nombre si es distinto

print("Dimensiones:", df.shape)
print("\nTipos de datos:")
print(df.dtypes)

print("\nPrimeras filas:")
print(df.head())

print("\nDescripción numérica:")
print(df.describe())

print("\nValores únicos por columna:")
for col in df.columns:
    print(col, "->", df[col].nunique(), "valores únicos")

# =========================
# 2. Calidad de datos
# =========================
print("\nValores faltantes por columna:")
print(df.isna().sum())

# Porcentaje de NA
na_percent = df.isna().mean() * 100
print("\nPorcentaje de NA por columna:")
print(na_percent)

# =========================
# 3. Variables demográficas
# =========================
dem_cols = ["Q1_Gender", "Q2_Living_Area", "Q3_Marital_Status"]

for col in dem_cols:
    print(f"\nFrecuencias de {col}:")
    print(df[col].value_counts(dropna=False))

    plt.figure()
    sns.countplot(x=col, data=df)
    plt.title(f"Distribución de {col}")
    plt.show()

# Opcional: mapear etiquetas
gender_map = {1: "Hombre", 2: "Mujer"}
area_map = {1: "Urbano", 2: "Rural"}
marital_map = {1: "No", 2: "Sí"}

df["Gender_label"] = df["Q1_Gender"].map(gender_map)
df["Area_label"] = df["Q2_Living_Area"].map(area_map)
df["Marital_label"] = df["Q3_Marital_Status"].map(marital_map)

# =========================
# 4. Ítems SMAQ1–SMAQ10
# =========================
item_cols = [col for col in df.columns if col.startswith("SMAQ")]

print("\nDescripción de los ítems SMAQ:")
print(df[item_cols].describe())

# Histograma por ítem
for col in item_cols:
    plt.figure()
    sns.histplot(df[col], bins=5, discrete=True)
    plt.title(f"Distribución de {col}")
    plt.xlabel("Respuesta Likert (1–5)")
    plt.ylabel("Frecuencia")
    plt.xticks([1, 2, 3, 4, 5])
    plt.show()

# Boxplot conjunto de ítems
plt.figure(figsize=(10, 5))
sns.boxplot(data=df[item_cols])
plt.title("Distribución de los ítems SMAQ1–SMAQ10")
plt.xlabel("Ítems")
plt.ylabel("Puntuación")
plt.show()

# =========================
# 5. Puntaje total de adicción
# =========================
# Si ya existe SMA_Scale_value en el dataset, puedes usarlo directamente.
if "SMA_Scale_value" in df.columns:
    df["SMA_total"] = df["SMA_Scale_value"]
else:
    df["SMA_total"] = df[item_cols].sum(axis=1)

print("\nDescripción del puntaje total (SMA_total):")
print(df["SMA_total"].describe())

plt.figure()
sns.histplot(df["SMA_total"], kde=True)
plt.title("Distribución del puntaje total de adicción (SMA_total)")
plt.xlabel("Puntaje total")
plt.ylabel("Frecuencia")
plt.show()

plt.figure()
sns.boxplot(x=df["SMA_total"])
plt.title("Boxplot del puntaje total de adicción")
plt.xlabel("SMA_total")
plt.show()

# =========================
# 6. Variable de clase (SMA_Scale)
# =========================
if "SMA_Scale" in df.columns:
    print("\nFrecuencias de SMA_Scale:")
    print(df["SMA_Scale"].value_counts())

    plt.figure()
    sns.countplot(x="SMA_Scale", data=df)
    plt.title("Distribución de niveles de adicción (SMA_Scale)")
    plt.xlabel("1=Baja, 2=Moderada, 3=Alta")
    plt.ylabel("Frecuencia")
    plt.show()
else:
    # Crear clase a partir de SMA_total si no existe
    def classify_sma(score):
        if score <= 20:
            return 1
        elif score <= 40:
            return 2
        else:
            return 3

    df["SMA_Scale"] = df["SMA_total"].apply(classify_sma)

    print("\nFrecuencias de SMA_Scale (creada):")
    print(df["SMA_Scale"].value_counts())

    plt.figure()
    sns.countplot(x="SMA_Scale", data=df)
    plt.title("Distribución de niveles de adicción (SMA_Scale)")
    plt.xlabel("1=Baja, 2=Moderada, 3=Alta")
    plt.ylabel("Frecuencia")
    plt.show()

# =========================
# 7. Matriz de correlación de ítems
# =========================
corr = df[item_cols].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Matriz de correlación entre ítems SMAQ")
plt.show()

# =========================
# 8. Puntaje total por grupos demográficos
# =========================
# Género
plt.figure()
sns.boxplot(x="Gender_label", y="SMA_total", data=df)
plt.title("Puntaje total de adicción por género")
plt.xlabel("Género")
plt.ylabel("SMA_total")
plt.show()

# Área
plt.figure()
sns.boxplot(x="Area_label", y="SMA_total", data=df)
plt.title("Puntaje total de adicción por área de residencia")
plt.xlabel("Área")
plt.ylabel("SMA_total")
plt.show()

# Estado civil
plt.figure()
sns.boxplot(x="Marital_label", y="SMA_total", data=df)
plt.title("Puntaje total de adicción por estado civil")
plt.xlabel("Estado civil")
plt.ylabel("SMA_total")
plt.show()

# =========================
# 9. Detección simple de outliers en SMA_total
# =========================
Q1 = df["SMA_total"].quantile(0.25)
Q3 = df["SMA_total"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["SMA_total"] < lower) | (df["SMA_total"] > upper)]
print("\nNúmero de posibles outliers en SMA_total:", outliers.shape[0])

# =========================
# 10. Resumen rápido de hallazgos (texto)
# =========================
print("\n--- RESUMEN RÁPIDO ---")
print("1) Distribución de ítems SMAQ revisada (histogramas y boxplots).")
print("2) Puntaje total SMA_total: forma de la distribución y posibles outliers.")
print("3) Desbalanceo de clases en SMA_Scale visualizado.")
print("4) Correlaciones entre ítems (heatmap) para ver redundancias o ítems débiles.")
print("5) Diferencias descriptivas de SMA_total por género, área y estado civil.")
```

El presente informe analiza los resultados obtenidos mediante la escala SMAQ (Smartphone Addiction Questionnaire) aplicada a una muestra amplia de participantes. La escala está compuesta por 10 ítems tipo Likert (1–5) que evalúan comportamientos asociados al uso problemático del smartphone.
El dataset incluye además variables sociodemográficas (género, área de residencia y estado civil), el puntaje total de la escala (SMA_Scale_value) y la clasificación final en tres niveles (SMA_Scale Class Level).

Variables sociodemográficas

Q1_Gender

Q2_Living_Area

Q3_Maritial_Status

Ítems de la escala SMAQ

SMAQ1 a SMAQ10

Puntaje total

SMA_Scale_value (rango observado: 10–50)

Clasificación final

1 = Bajo

2 = Moderado

3 = Alto

Ejemplos del documento:

“4,5,1,3,5,2,2,4,5,3 → 34 puntos → Nivel 2”
“3,2,1,1,1,1,1,1,1,1 → 13 puntos → Nivel 1”

Puntajes entre 10 y 18.

“SMA_Scale_value = 12 → Clase 1”

Puntajes entre 19 y 34.

Indica uso frecuente, pero no necesariamente problemático.

“SMA_Scale_value = 30 → Clase 2”

La mayoría de la muestra se concentra en el nivel moderado, seguido del nivel bajo. El nivel alto aparece con menor frecuencia, pero está claramente presente.
