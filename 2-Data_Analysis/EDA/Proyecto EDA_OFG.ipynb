```python
## checking outliers

numeric_cols = mobile.select_dtypes(include=['number', 'float', 'int']).columns.tolist()
z_outliers = pd.DataFrame(index=mobile.index)

for col in numeric_cols:
    mean = mobile[col].mean()
    std  = mobile[col].std(ddof=0)

    # avoid division by zero
    if std == 0:
        z_outliers[col] = False
        continue

    z_scores = (mobile[col] - mean) / std
    z_outliers[col] = (z_scores.abs() > 3)

z_outlier_summary = z_outliers.sum().sort_values(ascending=False)
z_outlier_summary
```
