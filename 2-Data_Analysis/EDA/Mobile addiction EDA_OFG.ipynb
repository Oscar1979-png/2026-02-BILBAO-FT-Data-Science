```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
##cargar el dataset original

mobile8 = pd.read_excel("ml_da_dataset.xlsx")
mobile8.head(10)
```

A partir del documento, las columnas son:
- Mutluluk → Felicidad
- Fomo → Fear of Missing Out
- Nevrotik → Neuroticismo
- Disdon → Deshonestidad / desorganización
- Acikfikir → Apertura mental
- Yumbas → Amabilidad / empatía
- Durust → Honestidad
- Mobba → Mobile-based addiction (subescala)
- Oyba → Online-youth behavior addiction
- Sosvirtol → Social virtual tolerance
- Sosvirco → Social virtual communication
- Sosvirpro → Social virtual productivity
- Sosvirinfo → Social virtual information
- Target → Variable objetivo (nivel de adicción)

```python
##renombramos las columnas

mobile9 = mobile8.rename(columns={
                        "Mutluluk": "Felicidad",
                        "Nevrotik": "Neuroticismo",
                        "Disdon": "Desorganización",
                        "Acikfikir": "Apertura mental",
                        "Yumbas": "Empatía",
                        "Durust": "Honestidad",
                        
})
inplace=True
mobile9.head()
```

Explicación de los resultados:
    
El dataset contiene 497 filas y 14 variables, todas numéricas.    
    

Estadísticas descriptivas (resumen interpretado)

Rasgos de personalidad

- Las escalas Sosvir* (tolerancia, comunicación, productividad, información) están entre 1.7 y 2.7.

Variable objetivo
- Target tiene media ≈ 0.92 → indica que la mayoría de los casos están etiquetados como “1”.

Correlaciones importantes del heatmap

Correlaciones fuertes dentro del grupo “Sosvir*”
Estas variables están muy correlacionadas entre sí:
- Sosvirtol
- Sosvirco
- Sosvirpro
- Sosvirinfo
Esto indica que forman un bloque coherente de comportamiento digital.

Mobba (adicción móvil) correlaciona moderadamente con:

Esto sugiere que mayor uso virtual → mayor adicción móvil.

FOMO correlaciona con:

Felicidad correlaciona negativamente con:

```python
plt.style.use('seaborn-v0_8')
mobile9.hist(bins=20, figsize=(14, 10), color='skyblue', edgecolor='black')
plt.tight_layout();
```

2. Boxplots

```python
plt.figure(figsize=(14, 8))
sns.boxplot(data=mobile9, palette='Set3')
plt.xticks(rotation=45)
plt.title('Boxplots de todas las variables')
plt.tight_layout()
```

```python
# 1. KDE plots for key variables
plt.style.use('seaborn-v0_8')
plt.figure(figsize=(12, 8))
for col in ['Felicidad', 'Fomo', 'Neuroticismo', 'Mobba', 'Sosvirinfo']:
    sns.kdeplot(mobile9[col], fill=True, alpha=0.4, label=col)
plt.title('KDE Plots de Variables Clave')
plt.legend();
```

1. KDE Plots — Distribuciones suaves.
Estos gráficos muestran la forma de la distribución de las variables clave:
- FOMO y Neuroticismo tienen distribuciones amplias → mucha variabilidad emocional.
- Felicidad se concentra en valores altos → población relativamente satisfecha.
- Mobba (adicción móvil) tiene un pico claro entre 1.5 y 2.5 → nivel moderado de adicción.
- Sosvirinfo muestra una distribución más plana → el uso informativo del móvil es muy diverso.

```python
# 2. Violin plots
plt.figure(figsize=(14, 8))
sns.violinplot(data=mobile9[['Felicidad', 'Fomo', 'Neuroticismo', 'Mobba', 'Sosvirinfo']], palette='Set2')
plt.title('Violin Plots de Variables Psicológicas y de Adicción')
plt.xticks(rotation=45);
```

2. Violin Plots — Distribución + densidad + outliers.

Los violines muestran:
- Neuroticismo y FOMO tienen colas largas → algunos usuarios muy altos en ansiedad digital.
- Felicidad es más compacto → menos variabilidad en felicidad.
- Mobba tiene una distribución bimodal ligera → dos perfiles de uso móvil.
- Sosvirinfo muestra dispersión amplia → distintos estilos de consumo informativo.

Las variables psicológicas tienen patrones distintos, lo que favorece la identificación de perfiles.

```python
# 3. Target vs variables (boxplots)

plt.figure(figsize=(14, 10))
for i, col in enumerate(['Felicidad', 'Fomo', 'Neuroticismo', 'Mobba', 'Sosvirinfo']):
    plt.subplot(2, 3, i + 1)
    sns.boxplot(x=mobile9['Target'], y=mobile9[col], palette='coolwarm', hue=1, legend=False)
    plt.title(f'{col} según Target')
plt.tight_layout()
```

3. Boxplots por Target — Comparación entre clases

Estos gráficos son especialmente útiles:

Target = 1 (usuarios con mayor riesgo) muestran:

- FOMO más alto
- Neuroticismo más alto
- Mobba más alto
- Sosvirinfo más alto


Target = 0 (usuarios de bajo riesgo) muestran:

- Felicidad más alto
- Menor variabilidad en uso digital

```python
# 4. Jointplot: FOMO vs Mobba
jp = sns.jointplot(data=mobile9, x='Fomo', y='Mobba', kind='hex', cmap='viridis')
jp.fig.suptitle('Jointplot: FOMO vs Mobba', y=1.02);
```

4. Pairplot — Relaciones visuales entre variables clave.

El pairplot muestra:

- Target = 1 (adicción) suele asociarse con:

- FOMO más alto
- Neuroticismo más alto
- Mobba más alto
- Sosvirinfo más alto
- Felicidad se separa visualmente: Los usuarios con mayor felicidad tienden a tener Target = 0.


```python
# 5. Pairplot extendido
pp = sns.pairplot(mobile9[['Felicidad', 'Fomo', 'Neuroticismo', 'Mobba', 'Sosvirinfo', 'Target']], hue='Target', diag_kind='kde')
pp.fig.suptitle('Pairplot de Variables Clave con Target', y=1.02);
```
