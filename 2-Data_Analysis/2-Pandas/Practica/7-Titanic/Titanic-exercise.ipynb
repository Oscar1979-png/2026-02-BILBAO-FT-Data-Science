27. Please, show only the rows using a mask with the following conditions: 
    - They are woman
    - From third class
    - Younger than 30
    - They survived
    
    ¿How many rows fulfill the condition?

```python
cond1 = titanic['gender']=='female'
cond2 = titanic['pclass']==3
cond3 = titanic['age']<30
cond4 = titanic['survived']==1

len(titanic[cond1 & cond2 & cond3 & cond4, :])
```

28. Now, show only the rows using `.loc` with the following conditions:
